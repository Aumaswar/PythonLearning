Numpy arrays and Basics


```python
import numpy as np
import time
```

creating array from list


```python
arr_1D = np.array([1,2,3,4,5])
print(arr_1D)

arr_2D = np.array([[1,2,3,4],[1,2,3,4]])
print(f"2D array: {arr_2D}")
```

List vs Numpy array

```python
py_list = [1,2,3]
print(f"Multiplication in list {py_list * 2}")
np_array = np.array([1,2,3])
print(f"Multiplication in array {np_array * 2}")


```

```python
start = time.time()
py_list = [i*2 for i in range(10000000)]
print(f"\n list opreation time: {time.time() - start}")
start = time.time()
np_array = np.arange(10000000) * 2
print(f"\n Array opreation time: {time.time() - start}")
```
